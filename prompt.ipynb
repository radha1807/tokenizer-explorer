what is USER PROMPT?

=> it is the actual task or the input given by the user.




What is ASSISTANT PROMPT?

=> it represents something the assistant has prevously said.


=> the model sees  the conversation history.

Why this is useful?

=> it can establish response format, conversation history, examples

However, don't confuse this with few-shot prompting. Few-shot prompting deliberately gives examples to teach a pattern; those examples can be represented using user/assistant message pairs or inside the prompt depending on the API.

what is ZERO SHOT PROMPTING?

=>  zero shot = no examples, you simply tell the model what to do.

=> for example 
Classify the following text as
POSITIVE, NEGATIVE, or NEUTRAL.

Text:
"The product works well."

When useful?

=> Use zero-shot when:
- task is simple
- expected behavior is obvious
- output format is straightforward
Google describes zero-shot as prompting without examples, while few-shot includes examples that demonstrate the desired pattern.

FEW SHOT PROMPTING

in FSP the model is given examples before asking it to perform the task



Why does it work?

=> the examples show the model:
- what the task means
- what the output should look like
- how detailed the answer should be
- how edge cases should be handled
Few-shot examples are especially useful for controlling format and behavior.


CHAIN of THOUGHT(CoT)

=> solving a problem through multiple reasoning steps rather than jumping directly to the answer.





ReACT PROMPTING 

=> ReAct = Reason + Act

=> The original ReAct research proposed interleaving reasoning and actions so an LLM can interact with external sources/environments instead of relying only on internal knowledge.

=> Instead of the model only thinking about a problem, it can:
Reason
  ↓
Take an action
  ↓
Observe result
  ↓
Reason again
  ↓
Take another action 



Where this is used?

- ai agents
- tool calling
- web search
- rag agents

ReAct is worth understanding, even though you dont need to master agent frameworks today.



XML PROMPTING

=> XML means using xml style tags to clearly seperate diff parts of your prompt.

- Instead of:

=> Analyze this document and summarize it. 
The document is about SPI communication...
Also don't include unnecessary details...

- use

=> <instruction>
Analyze the document and summarize the important findings.
</instruction>

<context>
The document describes SPI communication.
</context>

<requirements>
- Keep the answer concise.
- Focus on technical issues.
</requirements>



This becomes especially useful when prompts contain:
- instructions
- documents
- user-generated content
- examples
- multiple pieces of context

=> Anthropic specifically recommends XML tags for separating instructions, context, examples, and inputs in complex prompts

What is JSON Schema?
=> JSON Schema describes what the JSON output is allowed to look like.



- This basically tells the model/application:
- Output must contain:

error      → string
frequency  → number
severity   → string



what is STRUCTURED OUTPUTS?

=> SO take the previous idea further.

=> Instead of merely saying:
"Please return JSON."
- you define a schema that the model's output should conform to.

- conceptually

                LLM
                  ↓
          ┌───────────────┐
          │ JSON Schema   │
          │               │
          │ error: string │
          │ freq: number  │
          │ severity: str │
          └───────┬───────┘
                  ↓
        Structured response

- OpenAI's documentation distinguishes ordinary JSON mode from Structured Outputs: JSON mode aims for valid JSON, whereas Structured Outputs are designed to make the response conform to a supplied schema on supported configurations.
